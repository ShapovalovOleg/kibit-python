# Concatenation

```python
import numpy as np
ser1 = pd.Series(np.random.rand(4), index=[1, 2, 3, 4])
ser2 = pd.Series(np.random.rand(4), index=[5, 6, 7, 8])
```

```python
ser1
```

```python
ser2
```

```python
pd.concat([ser1, ser2])
```

```python
pd.concat([ser1, ser2], axis=1)
```

```python
pd.concat([ser1, ser2], axis=1, keys=['ser1', 'ser2'])
```

```python
pd.concat([ser1, ser2], keys=['Hello', 'World'])
```

```python
frame1 = pd.DataFrame(np.random.rand(9).reshape(3, 3),
                      index=[1, 2, 3], columns=['A', 'B', 'C'])
frame1
```

```python
ser1
```

# Pivoting

```python
frame1 = pd.DataFrame(
    np.arange(9).reshape(3, 3),
    index=['white', 'black', 'red'],
    columns=['ball', 'pen', 'pencil']
)
frame1
```

```python
tab1 = frame1.stack()
tab1
```

```python
original = tab1.unstack()
original
```

```python
tab1.unstack(level=0)
```

```python
longframe = pd.DataFrame({
    'color': ['white'] * 3 + ['red'] * 3 + ['black'] * 3,
    'item': ['ball', 'pen', 'mug'] * 3,
    'value': np.random.rand(9)
})
longframe
```

```python
wideframe = longframe.pivot(index='color', columns='item')
wideframe
```

```python
wideframe2 = longframe.pivot(index='item', columns='color')
wideframe2
```
