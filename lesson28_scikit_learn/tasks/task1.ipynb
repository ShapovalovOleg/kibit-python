```python
from sklearn import datasets
from sklearn.metrics import accuracy_score

wine = datasets.load_wine()
X = wine.data[:, [1, 9]]   # malic_acid, color_intensity
y = wine.target
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=50, random_state=42, shuffle=True
)
```

```python
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt

ks = [1, 2, 3, 5, 7]
acc = []
for k in ks:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    accuracy = knn.score(X_test, y_test)
    print(f"Accuracy for k={k}: {accuracy}")
    acc.append(accuracy)
```

```python
plt.figure(figsize=(7, 5))
scatter = plt.scatter(X_test[:, 0], X_test[:, 1], c=y_test, cmap="viridis", edgecolor="k", s=80)
plt.xlabel("Alcohol")
plt.ylabel("Flavanoids")
plt.title("Scatter тестових точок Wine dataset")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.bar(ks, acc, color='skyblue', edgecolor='k')
plt.xlabel("k (кількість сусідів)")
plt.ylabel("Accuracy")
plt.title("Точність KNN для різних k")
plt.show()
```
